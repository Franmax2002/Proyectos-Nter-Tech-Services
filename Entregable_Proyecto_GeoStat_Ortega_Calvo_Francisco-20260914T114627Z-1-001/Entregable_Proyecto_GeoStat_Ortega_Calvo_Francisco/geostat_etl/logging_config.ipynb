# `logging_config.ipynb` - Configuración centralizada del logging

Formato solicitado: `timestamp - nivel - mensaje`, con `filemode='w'` para obtener un **log limpio en cada ejecución**.

`configurar_logging()` debe llamarse **una única vez**, al principio de `main.ipynb` (después de `%run geostat_etl/config.ipynb`, ya que necesita `RUTA_LOG`), y devuelve el logger raíz del proyecto. El resto de módulos obtienen su logger con `logging.getLogger(__name__)`, que propaga los mensajes hacia el handler configurado aquí.

> **Nota sobre el idioma de los logs**: los mensajes que se escriben en el fichero de log (`logger.info` / `logger.warning` / `logger.error`) van **sin tildes** de forma deliberada, para evitar problemas de codificación al leer el log en distintos entornos/terminales (p. ej. `cp1252` en Windows). La documentación en markdown y los comentarios del código sí mantienen la ortografía normal con tildes.

```python
import logging
```

## Función de configuración

```python
def configurar_logging():
    """
    Configura el logging del proyecto con el formato exacto solicitado
    y devuelve el logger raiz del paquete ("geostat_etl"). Debe
    llamarse una unica vez, al principio de main.ipynb, antes de
    ejecutar cualquier otra funcion del proyecto.
    """
    logging.basicConfig(
        filename=RUTA_LOG,
        level=logging.INFO,
        format='%(asctime)s - %(levelname)s - %(message)s',
        datefmt='%Y-%m-%d %H:%M:%S',
        filemode='w',   # log limpio en cada ejecucion
        force=True,
    )
    return logging.getLogger("geostat_etl")
```

## Prueba rápida

Se configura el logging apuntando a `RUTA_LOG` (definida en `config.ipynb`, ya cargada en este namespace por el `%run` previo) y se escribe un mensaje de prueba, para confirmar que el fichero de log se crea y el formato es el correcto.

```python
logger = configurar_logging()
logger.info("Prueba de logging desde logging_config.ipynb")

with open(RUTA_LOG, encoding="utf-8") as f:
    print(f.read().strip().splitlines()[-1])
```
