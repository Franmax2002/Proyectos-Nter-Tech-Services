# `logging_config.ipynb` - Configuración centralizada del logging

Mismo patrón que en proyectos anteriores: formato `timestamp - nivel - mensaje`, `filemode='w'` para un log limpio en cada ejecución. Los mensajes de log van **sin tildes** a propósito (evitar problemas de codificación entre entornos); la documentación en markdown sí las mantiene.

```python
import logging
```

## Función de configuración

```python
def configurar_logging():
    """
    Configura el logging del proyecto con el formato exacto solicitado
    y devuelve el logger raiz ("geokw_etl"). Debe llamarse una unica vez,
    al principio de main.ipynb, antes de ejecutar cualquier otra funcion.
    """
    logging.basicConfig(
        filename=RUTA_LOG,
        level=logging.INFO,
        format='%(asctime)s - %(levelname)s - %(message)s',
        datefmt='%Y-%m-%d %H:%M:%S',
        filemode='w',
        force=True,
    )
    return logging.getLogger("geokw_etl")
```

## Prueba rápida

```python
logger = configurar_logging()
logger.info("Prueba de logging desde logging_config.ipynb")

with open(RUTA_LOG, encoding="utf-8") as f:
    print(f.read().strip().splitlines()[-1])
```
