# `logging_config.ipynb` - Configuracion del logger

Configura, una unica vez, el logger estandar de Python que usaran todos
los demas modulos del pipeline (extraccion, bronze, silver, gold, cli,
dashboard).

> **Decision de diseno**: se respeta la convencion de
> escribir en el log **sin tildes ni "n" con virgulilla**, para evitar
> problemas de encoding al abrir el fichero en distintos entornos
> (especialmente Windows con cp1252). Esta regla aplica solo a los
> mensajes de `logger.info/warning/error` y a los textos de excepciones
> que se registran; el resto del notebook (markdown, docstrings, prints
> para el usuario) mantiene la ortografia normal en espanol.

> Depende de: `config.ipynb` (usa `RUTA_LOG`).

```python
import os
if os.getcwd().endswith("climate_engine"):
    os.chdir("..")
%run climate_engine/config.ipynb
```

## Funcion principal

```python
import logging

logging.basicConfig(
    filename=str(RUTA_LOG),
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S",
    filemode="w",  # el log se sobreescribe en cada ejecucion
    force=True,
)
logger = logging.getLogger("climate_engine")


def separador_inicio(titulo):
    """Escribe en el log un separador visual y un titulo de inicio de fase.

    Se usa al comienzo de cada fase del pipeline (extraccion, bronze,
    silver, gold) para que el fichero de log sea facil de recorrer.
    """
    logger.info("=" * 70)
    logger.info(titulo)
    logger.info("=" * 70)
```

## Prueba rapida

Se escribe una entrada de prueba en el log y se relee el fichero para confirmar visualmente el formato exacto (fecha, nivel, mensaje sin acentos).

```python
separador_inicio("PRUEBA DE CONFIGURACION DE LOGGING - CLIMATE ENGINE")
logger.info("Prueba de log sin tildes: conexion, extraccion, validacion")
logger.warning("Prueba de log de aviso: fallo simulado de conexion a la API")

with open(RUTA_LOG, "r", encoding="utf-8") as f:
    contenido = f.read()

print(f"Fichero de log: {RUTA_LOG}")
print("--- Contenido actual del log ---")
print(contenido)

assert "conexion" in contenido and "conexión" not in contenido, (
    "El log no debe contener tildes en los mensajes registrados."
)
print("Validacion de formato de log: OK")
```
